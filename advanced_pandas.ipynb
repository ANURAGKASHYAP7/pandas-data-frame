```python
'''Q6 Multi-Level Grouping

Group the dataset by: Region and Category
Calculate the total sales for each combination.

What insights can you derive from this analysis?

'''
df.groupby(["Region","Category"])["Sales"].sum().sort_values(ascending=False)
```

```python
'''Q7 Pivot Table Creation

Create a pivot table showing:

Rows → Region

Columns → Product

Values → Total Sales

Which region sells the most laptops?
'''
pd.pivot_table(df,index="Region",columns="Product",values="Sales",aggfunc="sum")
```

**Q7(2)-East region sells the most laptops, with total sales of 55,568.**

```python
'''Q10 End-to-End Data Wrangling Task

Perform the following workflow:

Filter the dataset where Sales > 3000

Group data by Region and Product

Calculate total sales

Reshape the result into a pivot table

Explain the business insights obtained from this analysis.
'''
filtered_df = df[df["Sales"] > 3000]
grouped_sales = filtered_df.groupby(["Region", "Product"])["Sales"].sum()
```

```python
df.groupby(["Region","Product"])["Sales"].sum()
```

```python
pd.pivot_table(
    filtered_df,
    index="Region",
    columns="Product",
    values="Sales",
    aggfunc="sum"
)
```

```python
pivot_table = grouped_sales.unstack()
pivot_table
```

**Q10(5)Business Insight: The filtered data shows different sales patterns across regions and products. East has high sales for Laptop and Phone, while West has particularly high sales for Keyboard.
